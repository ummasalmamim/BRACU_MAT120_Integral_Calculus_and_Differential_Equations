```python
import sympy as smp
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = smp.symbols('x')
print(x/2)
display(x/2)
```

```python
x=smp.symbols('x')

f=smp.sin(x)
print("f(x)=")
display(f)

d=np.arange(-np.pi,np.pi)
print(d)

flam=smp.lambdify(x,f)
print(flam(d))

plt.plot(d,flam(d))
```

```python
x=smp.symbols('x')

f=smp.cos(x)
print("f(x)=")
display(f)

d=np.arange(-np.pi,np.pi,0.01)
print(d)

flam=smp.lambdify(x,f)
print(flam(d))

plt.plot(d,flam(d))
```

```python
f=smp.cos(x)
df=smp.diff(f,x)
print(df)
d=np.arange(-np.pi,np.pi,0.01)
#print(d)

flam=smp.lambdify(x,df)
#print(flam(d))

plt.plot(d,flam(d))
```

**Integration**

```python
fdx=smp.exp(-x)*(x**2)*smp.cos(x)
display(smp.Integral(fdx,x))
print("=")
f=smp.integrate(fdx,x)
display(f)
print("=")
display(smp.simplify(f))
```

```python
fdx=smp.exp(x)/smp.sqrt(smp.exp(x)+9)
display(smp.Integral(fdx,x))
print("=")
f=smp.integrate(fdx,x)
display(f)
print("=")
display(smp.simplify(f))
```

```python
fdx=smp.exp(-x)*(x**2)*smp.cos(x)
display(smp.Integral(fdx,(x,0,smp.ln(4))))
print("=")
f=smp.integrate(fdx,x)
display(f)
print("=")
display(smp.simplify(f))
```
