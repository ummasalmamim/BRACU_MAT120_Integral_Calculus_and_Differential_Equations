```python
import sympy as smp
import numpy  as np
import math
import matplotlib.pyplot as plt
from scipy.integrate import odeint
from IPython.display import display
```

```python
y=smp.Function("y")
x=smp.Function("x")
```

```python
x=smp.Symbol("x")
dy=smp.diff(y(x),x)
display(dy)
eq=dy+y(x)**2/x**2
display(eq)
sol2=smp.dsolve(eq)
display(sol2)
```
