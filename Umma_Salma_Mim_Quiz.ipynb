```python
from sympy import *
from numpy import *
from scipy.integrate import quad
import numpy as np
import sympy as smp
import math
import matplotlib.pyplot as plt
```

```python
x=smp.symbols("x")
```

```python
#Q1
fn=((smp.sin(x)*smp.cos(x))/(1-((1/5)*(x)**2)))
display(fn)

print("The Answer of the Derrivative:")
Temporary=smp.diff(fn,x)
display(Temporary)


h=lambdify(x,fn)
dh=lambdify(x,Temporary)

r=np.arange(-5,5,0.1)  #change this later
m=h(r)
dp=dh(r)

plt.plot(r,m,c="green")
plt.plot(r,dp,c="green")
plt.axhline(y=0,c="Gray")
plt.axvline(x=0,c="Gray")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
#Q2
#a)
def trapezoidal_rule(f,lower,upper,n):

    expression=smp.lambdify(x,f)

    gap=(upper-lower)/n
    sum=0.5*(expression(lower)+expression(upper))

    for i in range(1,n):
        data=lower+(i*gap)
        sum+=expression(data)

    return sum*gap


f=((smp.tan(x)**3)*ln(x))/x
display(smp.Integral(f,(x,1,50)))

print("Trapezoidal rule:")
print()
h=trapezoidal_rule(f,1,50,500)
display(h)

```
