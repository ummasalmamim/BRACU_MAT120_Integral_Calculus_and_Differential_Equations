#**Trapezoidal Rule**
For approximating definite integrals

```python
import sympy as smp
import numpy  as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
```

```python
def trapezoidal_rule(f,a,b,n):


    f=smp.lambdify(x,f)

    h=(b-a)/n
    sum=(f(a)+f(b))/2

    for i in range (1,n):
        x_i=a+i*h
        sum+=f(x_i)
    return(sum*h)

x=smp.symbols ('x')

f=x**2
a=0
b=10
n=100

print(smp.integrate(f,(x,a,b)).evalf())
print(trapezoidal_rule(f,a,b,n))
```

```python
y=smp.Function('y')
dy=smp.diff(y(x),x)
display(y(x))
display(dy)

OOE=dy+(y(x)**2)/x**2
display(OOE)
sol1=smp.dsolve(OOE,y(x))
sol2=smp.dsolve(OOE,y(x),ics={y(1):1/2})
print("Solution without intial condition:")
display(sol1)
print("Solution without intial condition:")
display(sol2)
```
